获取全部数据范例(首次更新)：

```python
import datetime
import pandas as pd
import api_demo
```

```python
###########获取当前日期###############
today=datetime.datetime.now()
```

```python
########获取历史年份数据#########
nav_his_year=await api_demo.get_fund_auth_nav_batch(2015,today.year)
########获取历史月份数据#########
if today.month >1:
    nav_his_month=await api_demo.get_fund_auth_nav_batch(today.year*100+1,today.year*100+today.month-1)
########获取历史日期数据#########
if today.day >1:
    nav_his_day=await api_demo.get_fund_auth_nav_batch(today.year*10000+today.month*100+1,today.year*10000+today.month*100+today.day-1)
######根据更新时间保留最新数据###############
nav_total=pd.concat([nav_his_year,nav_his_month,nav_his_day]).sort_values(by='update_time',ascending=False).groupby(['fund_id','statistic_date']).head(1)
```
